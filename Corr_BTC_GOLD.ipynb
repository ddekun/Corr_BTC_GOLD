# Статистическое сопоставление цены биткоина с ценами на золото.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
BTC = pd.read_csv('/content/drive/MyDrive/Geekbrains/prob_theory_and_math_stat/BTC.csv')
GOLD = pd.read_csv('/content/drive/MyDrive/Geekbrains/prob_theory_and_math_stat/gold.csv')
```

```python
BTC
```

```python
BTC.info()
```

```python
BTC.rename(columns = {'Откр.' : 'BTC open', 'Дата' : 'Date'}, inplace=True)
```

```python
BTC['BTC open'] = BTC['BTC open'].replace('[.]', '', regex=True).replace('[,]', '.', regex=True).astype(float)
```

```python
BTC['Date'] = BTC['Date'].replace('[.]', '-', regex=True)
```

```python
BTC_open = BTC[['Date', 'BTC open']].copy()
BTC_open.info()
```

```python
BTC_open['Date'] = BTC_open['Date'].str.split('-').str[::-1].str.join('-')
```

```python
BTC_open
```

```python
GOLD
```

```python
GOLD.rename(columns = {'Open' : 'GOLD open'}, inplace=True)
GOLD_open = GOLD[['Date', 'GOLD open']]
GOLD_open.head()
```

```python
res1 = BTC_open.merge(GOLD_open, on=["Date"])
res1.head()
```

```python
res1.info()
```

```python
res1[['BTC open', 'GOLD open',]].corr()
```

Вывод: Мы видим, что корреляция биткоина и золота высокая, это значит направления движения цены вверх или вниз чаще совпадают.

**Проверка не совпадающих дат**

```python
BTC_open['Date'].isin(GOLD_open['Date']).all()
GOLD_open['Date'].isin(BTC_open['Date']).all()
```

```python
(~BTC_open['Date'].isin(GOLD_open['Date'])).sum()
(~GOLD_open['Date'].isin(BTC_open['Date'])).sum()
```

```python
BTC_open[~BTC_open['Date'].isin(GOLD_open['Date'])].tail(20)
```

Вывод: часть дат не совпала, потому что у Биткоина и золота разный календарь торгов, золото не торгуется по выходным. А так же часть не совпала, потому что данные по золоту до 2022 года, а по Биткоину до 2023 года.
